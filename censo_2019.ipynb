```python
import pandas as pd
import numpy as np # Import numpy for np.nan check

# Read the excel file, specifying header=4 so the 5th row (index 4) becomes the first row of the DataFrame
df = pd.read_excel("censo 2019.xlsx", header=4)

# Set the column names to the values in the first row of the DataFrame, converting to string to avoid MultiIndex
df.columns = df.iloc[0].astype(str)

# Drop the first row as it has now become the column names
df = df[1:]

# Reset the index of the DataFrame for clean numbering
df = df.reset_index(drop=True)

# The first column name might be 'nan' (from the 'Unnamed: 0' column which had NaN in df.iloc[0,0])
# Let's drop this column if its name is 'nan' (string representation of NaN) as it seems to be an artifact.
if df.columns[0] == 'nan':
    df = df.drop(columns=df.columns[0], axis=1)

df.head()
```

```python
# Drop rows with any NaN values
df.dropna(inplace=True)

print(f"Shape of DataFrame after dropping rows with NaNs: {df.shape}")
display(df.head())
```

```python
print(df.columns.tolist())
```

```python
columns_to_drop = ['Código']
df = df.drop(columns=columns_to_drop, errors='ignore')
display(df.head())
```

```python
df.to_excel('censo 2019.xlsx', index=False)
print('DataFrame saved to censo 2019.xlsx')
```
